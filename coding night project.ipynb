```python
#### ======================================================
# SMART STUDENT MANAGEMENT SYSTEM
# File: smart_student_system.py
# Author: [Your Name]
# ======================================================

import os

FILE_NAME = "students.txt"

# ------------------------------------------------------
# Helper Functions
# ------------------------------------------------------

def validate_input(prompt, validation_func, error_message):
    """Generic input validator to avoid code repetition."""
    while True:
        value = input(prompt).strip()
        if validation_func(value):
            return value
        else:
            print("❌", error_message)

def load_students():
    """Read all student records from the file into a list of dicts."""
    students = []
    if not os.path.exists(FILE_NAME):
        return students
    with open(FILE_NAME, "r") as file:
        for line in file:
            if line.strip():
                parts = line.strip().split(",")
                if len(parts) == 5:
                    student = {
                        "id": parts[0],
                        "name": parts[1],
                        "age": parts[2],
                        "grade": parts[3],
                        "marks": parts[4],
                    }
                    students.append(student)
    return students

def save_students(students):
    """Write all students back to the file."""
    with open(FILE_NAME, "w") as file:
        for s in students:
            line = f"{s['id']},{s['name']},{s['age']},{s['grade']},{s['marks']}\n"
            file.write(line)

# ------------------------------------------------------
# CRUD Operations
# ------------------------------------------------------

def add_student():
    print("\n--- Add New Student ---")
    students = load_students()

    # Validation rules
    def valid_id(x):
        return x.isdigit() and x not in [s["id"] for s in students]

    def valid_name(x):
        return x.replace(" ", "").isalpha()

    def valid_age(x):
        return x.isdigit() and int(x) > 0

    def valid_grade(x):
        return len(x) == 1 and x.upper() in "ABCDEF"

    def valid_marks(x):
        return x.isdigit() and 0 <= int(x) <= 100

    sid = validate_input("Enter ID: ", valid_id, "Invalid or duplicate ID.")
    name = validate_input("Enter Name: ", valid_name, "Invalid name.")
    age = validate_input("Enter Age: ", valid_age, "Invalid age.")
    grade = validate_input("Enter Grade (A–F): ", valid_grade, "Invalid grade.")
    marks = validate_input("Enter Marks (0–100): ", valid_marks, "Invalid marks.")

    with open(FILE_NAME, "a") as f:
        f.write(f"{sid},{name},{age},{grade.upper()},{marks}\n")

    print("✅ Student added successfully!")

def view_students():
    print("\n--- All Student Records ---")
    students = load_students()
    if not students:
        print("No records found.")
        return
    print(f"{'ID':<6}{'Name':<20}{'Age':<6}{'Grade':<8}{'Marks':<6}")
    print("-" * 50)
    for s in students:
        print(f"{s['id']:<6}{s['name']:<20}{s['age']:<6}{s['grade']:<8}{s['marks']:<6}")
    print(f"\nTotal Students: {len(students)}")

def search_student():
    print("\n--- Search Student ---")
    students = load_students()
    if not students:
        print("No data available.")
        return
    query = input("Enter ID or Name to search: ").strip().lower()
    results = [s for s in students if query in s["id"].lower() or query in s["name"].lower()]
    if results:
        print(f"\nFound {len(results)} result(s):")
        for s in results:
            print(f"{s['id']} | {s['name']} | Age: {s['age']} | Grade: {s['grade']} | Marks: {s['marks']}")
    else:
        print("❌ No matching student found.")

def update_student():
    print("\n--- Update Student ---")
    students = load_students()
    if not students:
        print("No data available.")
        return
    sid = input("Enter ID of student to update: ").strip()
    for s in students:
        if s["id"] == sid:
            print(f"Editing {s['name']} (ID: {s['id']})")
            s["name"] = input(f"New Name ({s['name']}): ") or s["name"]
            s["age"] = input(f"New Age ({s['age']}): ") or s["age"]
            s["grade"] = input(f"New Grade ({s['grade']}): ") or s["grade"]
            s["marks"] = input(f"New Marks ({s['marks']}): ") or s["marks"]
            save_students(students)
            print("✅ Record updated successfully!")
            return
    print("❌ Student ID not found.")

def delete_student():
    print("\n--- Delete Student ---")
    students = load_students()
    if not students:
        print("No data available.")
        return
    sid = input("Enter ID of student to delete: ").strip()
    for s in students:
        if s["id"] == sid:
            confirm = input(f"Are you sure you want to delete {s['name']}? (y/n): ").lower()
            if confirm == "y":
                students.remove(s)
                save_students(students)
                print("✅ Record deleted successfully!")
            else:
                print("Deletion canceled.")
            return
    print("❌ Student ID not found.")

# ------------------------------------------------------
# Data Analysis
# ------------------------------------------------------

def analyze_data():
    print("\n--- Data Analysis ---")
    students = load_students()
    if not students:
        print("No records to analyze.")
        return

    marks_list = [int(s["marks"]) for s in students]
    avg = sum(marks_list) / len(marks_list)
    top = max(students, key=lambda s: int(s["marks"]))
    below_avg = [s for s in students if int(s["marks"]) < avg]
    high, low = max(marks_list), min(marks_list)

    # Custom Insights
    grade_counts = {}
    for s in students:
        grade_counts[s["grade"]] = grade_counts.get(s["grade"], 0) + 1

    print(f"Average Marks: {avg:.2f}")
    print(f"Top Performer: {top['name']} ({top['marks']})")
    print(f"Students Below Average: {len(below_avg)}")
    print(f"Highest Marks: {high} | Lowest: {low}")
    print("--- Grade Distribution ---")
    for g, count in grade_counts.items():
        print(f"Grade {g}: {count} student(s)")

# ------------------------------------------------------
# Main Menu
# ------------------------------------------------------

def main():
    while True:
        print("\n====== Smart Student Record Analyzer ======")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Analyze Data")
        print("7. Exit")

        choice = input("Enter your choice (1-7): ").strip()
        if choice == "1":
            add_student()
        elif choice == "2":
            view_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            update_student()
        elif choice == "5":
            delete_student()
        elif choice == "6":
            analyze_data()
        elif choice == "7":
            print("\nThank you for using Smart Student Management System!")
            break
        else:
            print("❌ Invalid choice. Please try again.")

# ------------------------------------------------------
# Entry Point
# ------------------------------------------------------

if __name__ == "__main__":
    main()
```
